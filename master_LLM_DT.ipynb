## Data Cleaning and Preprocessing

```python
# Create a copy of the Dataframe
df = df.copy()
```

### Creating a Pivot Table

```python
# Create a pivot table with both 'Antall lønnstakere' and 'Gjennomsnittlig månedslønn (kr)' in separate columns
df_pivot = df.pivot_table(
    index=['kjønn', 'alder', 'yrke', 'kvartal'],
    columns='statistikkvariabel',
    values='value'
).reset_index()

# Flatten the MultiIndex columns
df_pivot.columns = [col if isinstance(col, str) else col[1] for col in df_pivot.columns]
```

```python
# Extract the year from the 'kvartal' column (taking the first 4 characters as the year)
df_pivot['year'] = df_pivot['kvartal'].str[:4].astype(int)

# Group by year and calculate the median for each group
df = df_pivot.groupby(['kjønn', 'alder', 'yrke', 'year']).agg({
    'Antall lønnstakere': 'median',
    'Gjennomsnittlig månedslønn (kr)': 'median'
}).reset_index()
```

```python
# Rename the columns to match your desired names
df = df.rename(columns={
    'Antall lønnstakere': 'antall_lønnstakere',
    'Gjennomsnittlig månedslønn (kr)': 'gjennomsnittlig_månedslønn'
})
```

```python
# Remove the 'year' column as it's no longer needed
df = df.drop(columns='year')
```

```python
# Display the resulting DataFrame
df.head()
```

### Handling Missing Values

```python
# Check for missing values
df.isnull().sum()
```

```python
# Remove rows where both salary is missing and employee count is zero
df = df[~((df['antall_lønnstakere'] == 0) & (df['gjennomsnittlig_månedslønn'].isna()))].copy()

# Fill missing salary values with the overall median
df.loc[:, 'gjennomsnittlig_månedslønn'] = df['gjennomsnittlig_månedslønn'].fillna(df['gjennomsnittlig_månedslønn'].median())
```

```python
# Check again for missing values
df.isnull().sum()
```

```python
# Count how many gender entries exist per occupation
gender_counts_per_occupation = df.groupby(['yrke', 'kjønn']).size().unstack()

# Identify occupations missing either 'Kvinner' or 'Menn'
missing_genders = gender_counts_per_occupation[gender_counts_per_occupation.isna().any(axis=1)]
print("Occupations missing a gender:")
print(missing_genders)
```

```python
# Get unique occupations and genders
occupations = df['yrke'].unique()
genders = ['Kvinner', 'Menn']

# Build full index of expected (yrke, kjønn) pairs
full_index = pd.MultiIndex.from_product([occupations, genders], names=['yrke', 'kjønn'])

# Identify which (yrke, kjønn) pairs are missing
existing_index = pd.MultiIndex.from_frame(df[['yrke', 'kjønn']])
missing_index = full_index.difference(existing_index)

# For each missing (yrke, kjønn), create a row with 0s and NaNs but same alder as existing entry
new_rows = []
for yrke, kjønn in missing_index:
  ref_row = df[df['yrke'] == yrke].iloc[0].copy()
  new_rows.append({
      'kjønn': kjønn,
      'yrke': yrke,
      'alder': ref_row['alder'],
      'antall_lønnstakere': 0,
      'gjennomsnittlig_månedslønn': np.nan
    })

# Append the new rows to the original DataFrame
df = pd.concat([df, pd.DataFrame(new_rows)], ignore_index=True)

# Sort if needed
df = df.sort_values(by=['yrke', 'kjønn']).reset_index(drop=True)
```

## Most Gender-Imbalanced Occupations

# Baseline Modeling on Real-World Data

## Train Baseline Decision Tree

```python
# Train a baseline decision tree on real SSB data
df_baseline = df.copy()

# Encode features
label_encoders_baseline = {}
for col in ["yrke", "alder"]:
  le = LabelEncoder()
  df_baseline[col] = le.fit_transform(df_baseline[col])
  label_encoders_baseline[col] = le

# Encode gender as binary
df_baseline["kjønn_encoded"] = df_baseline["kjønn"].map({"Kvinner": 0, "Menn": 1})

# Feature matrix and target
X_baseline = df_baseline[["alder", "yrke", "gjennomsnittlig_månedslønn"]]
y_baseline = df_baseline["kjønn_encoded"]

# Train-test split
X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
  X_baseline, y_baseline, test_size=0.2, random_state=42
)

# Corresponding sample weights from 'antall_lønnstakere'
weights_train = df_baseline.loc[X_train_b.index, 'antall_lønnstakere']
```

## Shallow Tree (depth=5)

```python
# Train Shallow Decision Tree
clf_baseline = DecisionTreeClassifier(max_depth=5, random_state=42)
clf_baseline.fit(X_train_b, y_train_b, sample_weight=weights_train)

# Cross-validation accuracy
cv_scores = cross_val_score(clf_baseline, X_baseline, y_baseline, cv=5)
print("--- Shallow Decision Tree (Depth = 5) ---")
print(f"Cross-Validation Accuracy (mean ± std): {cv_scores.mean():.3f} ± {cv_scores.std():.3f}")

# Train/test set accuracy
print(f"Train Accuracy: {clf_baseline.score(X_train_b, y_train_b):.3f}")
print(f"Test Accuracy: {clf_baseline.score(X_test_b, y_test_b):.3f}")
```

```python
# Plot the decision tree
plt.figure(figsize=(32, 20))
plot_tree(
  clf_baseline,
  feature_names=X_baseline.columns,
  class_names=["Kvinner", "Menn"],
  filled=True,
  rounded=True,
)
plt.title("Shallow Decision Tree (depth = 5)")
plt.tight_layout()
plt.show()
```

## Simple Tree (depth=3)

```python
# Train a simple decision tree
clf_simple = DecisionTreeClassifier(max_depth=3, random_state=42)
clf_simple.fit(X_train_b, y_train_b)

# Cross-validation accuracy for simple model
cv_scores_simple = cross_val_score(clf_simple, X_baseline, y_baseline, cv=5)
print("--- Simple Decision Tree (Depth = 3) ---")
print(f"Cross-Validation Accuracy (mean ± std): {cv_scores_simple.mean():.3f} ± {cv_scores_simple.std():.3f}")

# Train/test set accuracy
print(f"Train Accuracy: {clf_simple.score(X_train_b, y_train_b):.3f}")
print(f"Test Accuracy: {clf_simple.score(X_test_b, y_test_b):.3f}")
```

```python
# Plot the simple decision tree
plt.figure(figsize=(32, 20))
plot_tree(
  clf_simple,
  feature_names=X_baseline.columns,
  class_names=["Kvinner", "Menn"],
  filled=True,
  rounded=True,
  fontsize=9
)
plt.title("Simple Decision Tree (Depth = 3)")
plt.show()
```

## Generate LLM-Based Occupational Predictions

```python
# Load dataset
df_masked = pd.read_csv("masked_dataset.csv")

# Pronouns
male_pronouns = ["Han", "han"]
female_pronouns = ["Hun", "hun"]

# Family terms
male_family_terms = ["Faren", "faren", "Onkelen", "onkelen", "Broren", "broren", "Fetteren", "fetteren"]
female_family_terms = ["Moren", "moren", "Tanten", "tanten", "Søsteren", "søsteren", "Kusinen", "kusinen"]

# Gendered nouns
male_nouns = ["Mannen", "mannen", "Gutten", "gutten"]
female_nouns = ["Kvinnen", "kvinnen", "Jenten", "jenten"]

# Category dictionary used for counting
categories = {
    "Mann": male_pronouns + male_family_terms + male_nouns,
    "Kvinne": female_pronouns + female_family_terms + female_nouns
}

male_terms = male_pronouns + male_family_terms + male_nouns
female_terms = female_pronouns + female_family_terms + female_nouns
possible_candidates = male_terms + female_terms

# Enable dropout layers during inference
def enable_mc_dropout(model):
    for module in model.modules():
        if isinstance(module, torch.nn.Dropout):
            module.train()

# Common fill-mask function that accepts model/tokenizer
def fill_mask(sentences, model, tokenizer, batch_size=16):
  all_outputs = []

  for i in tqdm(range(0, len(sentences), batch_size)):
    batch = sentences[i:i + batch_size]
    inputs = tokenizer(batch.tolist(), return_tensors="pt", padding=True, truncation=True, max_length=512).to(device)

    mask_id = tokenizer.convert_tokens_to_ids("[MASK]")
    mask_token_indices = (inputs.input_ids == mask_id).nonzero(as_tuple=True)

    with torch.no_grad():
      outputs = model(**inputs)
      logits = outputs.logits

    for j in range(len(batch)):
      try:
        mask_idx = mask_token_indices[1][mask_token_indices[0] == j].item()
      except:
        all_outputs.append((batch.iloc[j], "[MASK]"))  # fallback
        continue

      mask_logits = torch.softmax(logits[j, mask_idx, :], dim=0)

      candidate_scores = {}
      for candidate in possible_candidates:
        tokens = tokenizer.tokenize(candidate)
        if not tokens:
            continue
        candidate_id = tokenizer.convert_tokens_to_ids(tokens[0])
        candidate_scores[candidate] = mask_logits[candidate_id].item()

      best_candidate = max(candidate_scores, key=candidate_scores.get)
      filled = batch.iloc[j].replace(tokenizer.mask_token, best_candidate)
      all_outputs.append((filled, best_candidate))

  return all_outputs

# Repeat each row n times for MC Dropout
n_repeats = 3
df_masked = pd.concat([df_masked]*n_repeats, ignore_index=True)
```

# Surrogate Decision Tree Analysis

## Encoding Features

```python
# Feature columns used to predict gender
feature_cols = ["Jobb", "Alder", "Utdanning", "Månedslønn"]

# Encode categorical variables
label_encoders = {}
categorical_columns = ["Jobb", "Alder", "Utdanning", "Månedslønn"]

for col in categorical_columns:
  le = LabelEncoder()
  df_masked[col] = le.fit_transform(df_masked[col])
  label_encoders[col] = le

# Add decoded columns using inverse transform
for col in categorical_columns:
  le = label_encoders[col]
  df_masked[f"{col}_decoded"] = le.inverse_transform(df_masked[col])
```

## Train Shallow Decision Trees (depth=5)

```python
# Dictionary to store each trained tree
trained_trees_shallow = {}
trained_X_shallow = {}

print("--- Shallow Decision Trees (Depth=5) ---")

# Loop through each LLM model prediction
for model_col in model_cols:
  # Prepare data
  y = df_masked[model_col].apply(assign_gender)
  mask = y.isin(["Mann", "Kvinne"])
  X_filtered = df_masked.loc[mask, feature_cols]
  y_filtered = y[mask]

  # Train-test split
  X_train, X_test, y_train, y_test = train_test_split(X_filtered, y_filtered, test_size=0.2, random_state=42)

  # Train Decision Tree
  clf = DecisionTreeClassifier(max_depth=5, random_state=42)
  clf.fit(X_train, y_train)

  # Cross-validation score
  scores = cross_val_score(clf, X_filtered, y_filtered, cv=5)
  print(f"Cross-Validation Accuracy (mean ± std): {scores.mean():.3f} ± {scores.std():.3f}")

  # Train/Test set accuracy
  print(f"Train Accuracy: {clf.score(X_train, y_train):.3f}")
  print(f"Test Accuracy: {clf.score(X_test, y_test):.3f}")


  # Save trained Decision Tree
  trained_trees_shallow[model_col] = clf
  trained_X_shallow[model_col] = X_train

  # Plot Decision Tree
  plt.figure(figsize=(16,8))
  plot_tree(clf, feature_names=feature_cols, class_names=["Kvinne", "Mann"], filled=True, rounded=True)
  plt.title(f"Shallow Decision Tree for {model_col}")
  plt.show()
```

## Equal Opportunity Difference

```python
# High-salary classification disparity
def get_salary_midpoint(salary_str):
  try:
    # Remove commas and use regex to get digits
    clean = re.sub(r"[^\d–-]", "", salary_str.replace(",", ""))
    parts = re.split(r"[–-]", clean)  # handle both hyphen and en dash
    if len(parts) == 2:
      lower, upper = map(int, parts)
      return (lower + upper) // 2
    elif len(parts) == 1:
      return int(parts[0])
  except:
      pass
  return 0

# Loop through all models
for model_col in model_cols:
  print(f"\nEqual Opportunity Difference for {model_col}")

  df_model = df_masked.copy()

  # Decode 'Månedslønn'
  df_model['Månedslønn_Decoded'] = label_encoders['Månedslønn'].inverse_transform(df_model['Månedslønn'])

  # Calculate Salary Midpoints
  df_model['Salary_Midpoint'] = df_model['Månedslønn_Decoded'].apply(get_salary_midpoint)

  # Define High Paying Job: Salary > 50,000 NOK
  df_model['High_Paying_Job'] = df_model['Salary_Midpoint'].apply(lambda x: 1 if x > 50000 else 0)

  # Map LLM predictions to gender (0 = Mann, 1 = Kvinne)
  df_model['Predicted_Gender'] = df_model[model_col].apply(assign_gender)
  gender_mapping = {"Mann": 0, "Kvinne": 1}
  df_model['Predicted_Gender'] = df_model['Predicted_Gender'].map(gender_mapping)

  # High paying job classified by gender
  high_salary_male = df_model[(df_model['Predicted_Gender'] == 0) & (df_model['High_Paying_Job'] == 1)]
  high_salary_female = df_model[(df_model['Predicted_Gender'] == 1) & (df_model['High_Paying_Job'] == 1)]

  # Total classified males/females
  total_male = len(df_model[df_model['Predicted_Gender'] == 0])
  total_female = len(df_model[df_model['Predicted_Gender'] == 1])

  # Proportions
  male_pct = len(high_salary_male) / total_male if total_male else 0
  female_pct = len(high_salary_female) / total_female if total_female else 0

  # Equal Opportunity Difference
  diff = female_pct - male_pct
  print(f"Equal Opportunity Difference: {diff:.4f}")

  # Optional: Extra details
  print(f"High-salary classified males: {len(high_salary_male)} / {total_male}")
  print(f"High-salary classified females: {len(high_salary_female)} / {total_female}")

  # Update existing row
  for row in evaluation_summary:
    if row["Model"] == model_col:
      row["Equal Opportunity Diff"] = diff
      break
```

# Embedding-Level Bias

## Log-Probability Bias Score (LPBS)

```python
def compute_lpbs_from_mask(model, tokenizer, df_masked, male_terms, female_terms, batch_size=16):
  model.eval()
  model.to(device)

  mask_token = tokenizer.mask_token
  sentences = df_masked["Setning"].tolist()
  bias_scores = []

  for i in tqdm(range(0, len(sentences), batch_size), desc="Computing LPBS"):
    batch = sentences[i:i + batch_size]
    inputs = tokenizer(batch, return_tensors="pt", padding=True, truncation=True, max_length=512).to(device)

    mask_token_id = tokenizer.convert_tokens_to_ids(mask_token)
    mask_positions = (inputs.input_ids == mask_token_id).nonzero(as_tuple=False)

    with torch.no_grad():
      logits = model(**inputs).logits
      log_probs = torch.log_softmax(logits, dim=-1)

    for (batch_idx, pos) in mask_positions:
      male_scores = []
      female_scores = []

      for male_term, female_term in zip(male_terms, female_terms):
        male_id = tokenizer.convert_tokens_to_ids(tokenizer.tokenize(male_term)[0])
        female_id = tokenizer.convert_tokens_to_ids(tokenizer.tokenize(female_term)[0])

        male_scores.append(log_probs[batch_idx, pos, male_id].item())
        female_scores.append(log_probs[batch_idx, pos, female_id].item())

      score = np.mean(male_scores) - np.mean(female_scores)
      bias_scores.append(score)

  return bias_scores
```

```python
model_tokenizer_map = {
  "Predicted_Token_nb_bert_base": (model_nb, tokenizer_nb),
  "Predicted_Token_norbert_base": (model_norbert_base, tokenizer_norbert_base),
  "Predicted_Token_norbert_large": (model_norbert_large, tokenizer_norbert_large),
  "Predicted_Token_bert_multilingual_cased": (model_bert_multilingual_cased, tokenizer_bert_multilingual_cased),
}

for model_col in model_cols:
  print(f"\nCalculating LPBS for {model_col}...")
  model, tokenizer = model_tokenizer_map[model_col]

  lpbs_scores = compute_lpbs_from_mask(model, tokenizer, df_masked, male_terms, female_terms)

  lpbs_mean = np.mean(lpbs_scores)
  lpbs_std = np.std(lpbs_scores)

  print(f"→ LPBS Mean: {lpbs_mean:.4f}, Std: {lpbs_std:.4f}")

  # Update evaluation_summary
  for row in evaluation_summary:
    if row["Model"] == model_col:
      row["LPBS Mean"] = lpbs_mean
      row["LPBS Std"] = lpbs_std
      break
```

## Word Embedding Association Test (WEAT)

```python
def get_embedding(word, tokenizer, model):
  token_ids = tokenizer.encode(word, add_special_tokens=False)
  if not token_ids:
    return None
  with torch.no_grad():
    embeddings = model.get_input_embeddings()(torch.tensor(token_ids).to(model.device))
  return embeddings.mean(dim=0).cpu().numpy()

def cosine_similarity(a, b):
  return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

def weat_score_weighted(gender_A, gender_B, occupations, weights, tokenizer, model):
  vec_A = [get_embedding(w, tokenizer, model) for w in gender_A if get_embedding(w, tokenizer, model) is not None]
  vec_B = [get_embedding(w, tokenizer, model) for w in gender_B if get_embedding(w, tokenizer, model) is not None]

  mean_A = np.mean(vec_A, axis=0)
  mean_B = np.mean(vec_B, axis=0)

  diffs = []
  used_weights = []

  for occ, weight in zip(occupations, weights):
    occ_vec = get_embedding(occ, tokenizer, model)
    if occ_vec is None:
        continue
    sim_A = cosine_similarity(occ_vec, mean_A)
    sim_B = cosine_similarity(occ_vec, mean_B)
    score = sim_A - sim_B
    diffs.append(score * weight)
    used_weights.append(weight)

  if len(used_weights) == 0:
    return None
  return np.sum(diffs) / np.sum(used_weights)
```

```python
# Clean occupation labels
df_real["yrke_clean"] = df_real["yrke"].astype(str).str.lower().str.strip()
df_real_filtered = df_real[df_real["yrke_clean"].notna() & df_real["antall_lønnstakere"].notna()]

occupations = df_real_filtered["yrke_clean"].tolist()
weights = df_real_filtered["antall_lønnstakere"].tolist()

# Results
weat_results = []

# Loop through all models
for model_col, (model, tokenizer) in model_tokenizer_map.items():
  print(f"Calculating WEAT for: {model_col}")
  model.eval()
  model.to(device)

  score = weat_score_weighted(
    gender_A=male_terms,
    gender_B=female_terms,
    occupations=occupations,
    weights=weights,
    tokenizer=tokenizer,
    model=model
  )

  weat_results.append({
    "Model": model_col,
    "WEAT Score": round(score, 4) if score is not None else "N/A"
  })

  # Optional: update evaluation_summary
  for row in evaluation_summary:
    if row["Model"] == model_col:
      row["WEAT Score"] = score
      break

# Display as DataFrame
pd.DataFrame(weat_results)
```
